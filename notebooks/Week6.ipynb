> ### Data prep
>
> A key step is to set up the data right. I recommend the following workflow:
>
> 1. Take the data for the period you want to plot, group it by hour-of-the-day for each of your Personal Focus Crimes. We'll use the entire period by default, but you may use another period if you'd like.
> 2. We want to compare *shapes* of distributions across crime types, not absolute numbers, so we'll work on **normalized data**:
>    * To normalize within a crime category, divide the count for each hour by the total number of that crime type. (For example, in the `Drug Offense` category, take the count of drug offenses in the midnight-to-1am hour and divide by the total number of drug offenses across all hours, etc)
>    * Note that when you normalize the data, you can interpret your data as a probability distribution, where each hour shows the probability that a crime will happen at that time.
>    * Your life will be easiest if you organize your data in **long format** — a DataFrame with columns like `Hour`, `CrimeType`, and `NormalizedCount`. If you have it in wide format (one column per crime type), look into [`pd.melt()`](https://pandas.pydata.org/docs/reference/api/pandas.melt.html) to reshape it.
>
> ### Creating the plot
>
> You'll use `px.bar` from Plotly Express — the same library you used for choropleth maps in Week 5. A few things to think about:
>
> * With many crime types, placing bars side-by-side for each hour produces tiny, unreadable bars. Look into Plotly's different [`barmode`](https://plotly.com/python/bar-charts/) options for a better approach. Adding some transparency will help when bars overlap.
> * The plot is more useful if it starts "empty" (all data/traces hidden) so the user can build up comparisons by clicking legend items one at a time. Look into Plotly's `visible` trace property to achieve this.
> * If all traces start hidden, Plotly has no data to auto-scale from, so your axes will not be meaningful. You'll need to set `range_x` and `range_y` explicitly. Think about what sensible values would be for 24 hours of data with normalized counts.
> * Place the legend where it doesn't obscure the data or the axis tick values. <font color=gray>It's OK to ask your LLM for help with legend placement.</font>

````python
# Prep the data
# Use complete years only (2003–2025)
df["Incident Date"] = pd.to_datetime(df["Incident Date"])
df_complete = df[df["Incident Date"].dt.year <= 2025].copy()

# Extract hour from Incident Time (format "HH:MM")
df_complete["Hour"] = df_complete["Incident Time"].str[:2].astype(int)

# All Unified Categories in the merged dataset are our Personal Focus Crimes
focus_crimes = sorted(df_complete["Unified Category"].dropna().unique())

# Group by Hour and CrimeType, count incidents
hourly = (
    df_complete[df_complete["Unified Category"].isin(focus_crimes)]
    .groupby(["Hour", "Unified Category"])
    .size()
    .reset_index(name="Count")
    .rename(columns={"Unified Category": "CrimeType"})
)

# Normalize within each crime type (count / total for that crime)
hourly["NormalizedCount"] = hourly["Count"] / hourly.groupby("CrimeType")[
    "Count"
].transform("sum")
````

````python
# Exercise 2.1
import plotly.express as px

# Create the plot
fig = px.bar(
    hourly,
    x="Hour",
    y="NormalizedCount",
    color="CrimeType",
    barmode="overlay",
    opacity=0.6,
    title="Normalized Hourly Distribution of Personal Focus Crimes — SF (2003–2025)",
    labels={
        "NormalizedCount": "Probability",
        "Hour": "Hour of Day",
        "CrimeType": "Crime Type",
    },
    range_x=[-0.5, 23.5],
    range_y=[0, 0.12],
)

# Start all traces hidden so user builds up comparisons by clicking legend items
fig.update_traces(visible="legendonly")

fig.update_layout(
    xaxis=dict(tickmode="linear", tick0=0, dtick=1),
    legend=dict(
        orientation="v",
        x=1.02,
        y=1,
        xanchor="left",
        yanchor="top",
        title="Crime Type",
    ),
    bargap=0.05,
)

fig.show()
````

> *Exercise 2.2*: Animated crime patterns.
>
> One of Plotly's powerful features is built-in animation. Let's use it to show how hourly crime patterns have changed over time.
>
> * Take the concept of normalized hourly distributions, but now compute one **separately for each year** (still for each of your Personal Focus Crimes).
>    * Take a look at the yearly counts. If some crime-types drop precipitously at some point, it might create noisy plots.
> * Create an animated **line chart** using Plotly Express's `animation_frame` parameter. Each frame should show one year's hourly distribution. Here's a sense of how to build the animation:
>
> ```python
> fig = px.line(your_dataframe, 
>              x='Hour', y='NormalizedCount', color='CrimeType',
>              animation_frame='Year',

> ```
>
> * **Important**: Set `range_y` to fix the y-axis range across all frames. Without this, Plotly will auto-scale each frame independently, which makes it difficult/impossible to compare across years. (This is a real-world pitfall — animated charts that silently re-scale their axes can be deeply misleading. Remember the perceptual errors discussion from Week 5?)
> * With 9+ crime types on screen at once, the plot can get crowded. Consider starting with only one or two traces visible and letting the user toggle the rest on via the legend.
````python
# Exercise 2.2 — Animated crime patterns

# Build per-year normalized hourly distributions
df_complete["Year"] = df_complete["Incident Date"].dt.year

hourly_yearly = (
    df_complete[df_complete["Unified Category"].isin(focus_crimes)]
    .groupby(["Year", "Hour", "Unified Category"])
    .size()
    .reset_index(name="Count")
    .rename(columns={"Unified Category": "CrimeType"})
)

# Normalize within each (Year, CrimeType) pair
hourly_yearly["NormalizedCount"] = hourly_yearly["Count"] / hourly_yearly.groupby(
    ["Year", "CrimeType"]
)["Count"].transform("sum")

# Sort so animation frames are in chronological order
hourly_yearly = hourly_yearly.sort_values(["Year", "CrimeType", "Hour"])

# Only hide on fig.data (initial render) — NOT in frames.
# If frames carry a 'visible' key they override user toggles on every frame transition.
first_crimes = sorted(hourly_yearly["CrimeType"].unique())[:2]


def hide_initial(fig, first_crimes):
    for trace in fig.data:
        if trace["name"] not in first_crimes:
            trace["visible"] = "legendonly"


# ── Animated line chart ───────────────────────────────────────────────────────
fig_line = px.line(
    hourly_yearly,
    x="Hour",
    y="NormalizedCount",
    color="CrimeType",
    animation_frame="Year",
    title="Hourly Crime Distribution by Year — SF Personal Focus Crimes (animated lines)",
    labels={
        "NormalizedCount": "Probability",
        "Hour": "Hour of Day",
        "CrimeType": "Crime Type",
    },
    range_x=[-0.5, 23.5],
    range_y=[0, 0.20],
)

hide_initial(fig_line, first_crimes)

fig_line.update_layout(
    xaxis=dict(tickmode="linear", tick0=0, dtick=1),
    legend=dict(x=1.02, y=1, xanchor="left", yanchor="top", title="Crime Type"),
)
fig_line.show()

# ── Animated bar chart ────────────────────────────────────────────────────────
fig_bar = px.bar(
    hourly_yearly,
    x="Hour",
    y="NormalizedCount",
    color="CrimeType",
    animation_frame="Year",
    barmode="group",
    title="Hourly Crime Distribution by Year — SF Personal Focus Crimes (animated bars)",
    labels={
        "NormalizedCount": "Probability",
        "Hour": "Hour of Day",
        "CrimeType": "Crime Type",
    },
    range_x=[-0.5, 23.5],
    range_y=[0, 0.20],
)

hide_initial(fig_bar, first_crimes)

fig_bar.update_layout(
    xaxis=dict(tickmode="linear", tick0=0, dtick=1),
    legend=dict(x=1.02, y=1, xanchor="left", yanchor="top", title="Crime Type"),
)
fig_bar.show()
````

**Favourite visualization:** The *Budget Forecasts, Compared With Reality* piece from the New York Times. **Exercise 3.2 — Answers**

**Placing our Part 2 visualizations on the author-driven / reader-driven spectrum:**

Both visualizations sit toward the *reader-driven* end. Exercise 2.1 (the overlaid bar chart) lets the user toggle crime types on/off via the legend with no prescribed order — it is essentially a visual query tool. Exercise 2.2 (the animated chart) adds a temporal axis via animation, but the user still chooses which traces to display and can scrub the timeline freely.

To push them toward the author-driven end, we could: (1) add annotations highlighting specific patterns (e.g. an arrow pointing to the midnight spike in drug offenses with explanatory text), (2) use a slideshow structure that steps the viewer through a curated sequence of years rather than letting them scrub freely, and (3) start with only one crime type visible and progressively reveal others with captions explaining what to compare.

**Two audiences:**

- **SF Board of Supervisors:** An *interactive slideshow* (Segel & Heer genre) would work well. Politicians need a guided narrative with a clear takeaway — "here is what changed, here is why it matters, here is what you can do." The slideshow structure lets us control the sequence (author-driven opening) while still offering interactive drill-down for supervisors who want to explore their own district. The martini glass model fits: narrate the key trends first, then open to exploration.

- **Peer-reviewed journal:** A narrative genre does *not* fit well here because the output is a static PDF figure — no interactivity is possible. Instead, the best approach is a *partitioned poster* or *annotated chart* style: small multiples (one subplot per year or per crime type) with careful annotations. The strength of small multiples is that they support comparison without animation, and they are reproducible and printable. The trade-off is density — you lose the ability to let the reader explore, but you gain precision and permanence.

**Critical reflection on policy use:**

If a city official used our animated hourly distributions to argue for increased late-night patrols in specific neighborhoods, they would be assuming that

(1) the *recorded* crime distribution reflects *actual* crime

(2) the temporal pattern is stable enough to predict future incidents. Both assumptions are problematic. As Richardson et al. argue, crime data reflects *where and when police choose to look*, not where crime actually happens. Deploying more officers at midnight in areas that already show high arrest counts will generate *more* arrests there — confirming the pattern in next year's data and justifying yet more deployment.

This is the same feedback loop we discussed in Week 5: the data is not a neutral measurement of reality, it is partly a product of the policing decisions it is used to justify. Our animated visualization, by making temporal hotspots vivid and seemingly objective, could make this feedback loop harder to see rather than easier.
